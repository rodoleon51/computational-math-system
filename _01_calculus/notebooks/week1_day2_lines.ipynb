### 1. Mathematical Background

**Distance formula**
d = sqrt( (x2 - x1)^2 + (y2 - y1)^2 )

**Midpoint formula**
M = ( (x1 + x2)/2 , (y1 + y2)/2 )

**Slope formula**
m = (y2 - y1) / (x2 - x1)
**Point–slope form of a line**
y - y1 = m(x - x1)

**Slope–intercept form of a line**
y = mx + b

**Distance from a point to a line**
Ax + By + C = 0

d = |Ax1 + By1 + C| / sqrt(A^2 + B^2)

**Quadrants**
- Q1: x>0, y>0
- Q2: x<0, y>0
- Q3: x<0, y<0
- Q4: x>0, y<0

```python
# Ch 18 Determine k so that points are the verticies of a right triangle.
# Python code to calculate k using slopes for a right angle at B

# Points
A = (7, 3)
B = (-1, 0)
C_y = -2  # y-coordinate of C is fixed

# Slope of AB
m_AB = (A[1] - B[1]) / (A[0] - B[0])  # (3 - 0) / (7 - (-1)) = 3/8
print(f"Slope of AB: {m_AB}")  # 0.375

# For perpendicularity: m_AB * m_CB = -1
# So m_CB = -1 / m_AB
m_CB_required = -1 / m_AB
print(f"Required slope of CB: {m_CB_required}")  # -8/3 ≈ -2.666...

k = -2 / m_CB_required - 1

print(f"Calculated k = {k}")
print(f"Exact fraction: k = -1/4")

# Alternative one-liner calculation
k_direct = -2 / (-1 / (3 / 8)) - 1
print(f"One-liner k: {k_direct}")
```

```python
# Ch3 20. Do points (10, 2) and (12, 3) lie on line 2x +5y = 10?
x, y = Symbol("x"), Symbol("y")
line_eq = 2*x + 5*y - 10
point1 = (10, 2)
point2 = (12, 3)
val1 = line_eq.subs({x: point1[0], y: point1[1]})
val2 = line_eq.subs({x: point2[0], y: point2[1]})
print(f"Value of 2*x + 5*y - 10 at point {point1}: {val1} (should be 0)")
print(f"Value of 2*x + 5*y - 10 at point {point2}: {val2} (should be 0)")
if val1 == 0 and val2 == 0:
    print("Both points lie on the line.")
else:
    print("At least one point does not lie on the line.")
```
