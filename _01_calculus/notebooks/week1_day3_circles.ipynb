## General Equation of a Circle

The general equation of a circle can be expressed in two forms:

**Standard Form (Center-Radius Form):**
$$(x - h)^2 + (y - k)^2 = r^2$$

where:
- $(h, k)$ is the center of the circle
- $r$ is the radius

**General Form:**
$$x^2 + y^2 + Dx + Ey + F = 0$$

where $D$, $E$, and $F$ are constants.

The two forms are related by:
- $h = -\frac{D}{2}$
- $k = -\frac{E}{2}$
- $r = \sqrt{h^2 + k^2 - F}$

```python
# 1a. find center and radius of the circle given by the equation
# 2x^2 + 2y^2 - 4x + y + 1 = 0
# Divide by 2 to reduce x and y coeffs to 1, then identify D, E, F
D = Rational(-2, 1)
E = Rational(1, 2)
F = Rational(1, 2)

h, k, r, r2 = general_to_center_radius_rational(D, E, F)

# Print solution if radius is real, else indicate no real solution
if r.is_real is True:
    print(f"Center: ({h}, {k}), Radius: {r}")
else:
    # Fallback check via r^2 in case r.is_real is None
    r2 = sp.simplify(r**2)
    if (r2.is_real is True) and (r2.is_nonnegative is True):
        print(f"Center: ({h}, {k}), Radius: {r}")
    else:
        print("the equation has no real solution")
```

```python
# 14. Find the standard equation of the circle that passes through the points (8, -5) and (-1, 4)
#     and has its center on the line 2x + 3y = 3.

import sympy as sp
x, y, h, k = sp.symbols("x y h k")

p0 = (sp.Integer(8), sp.Integer(-5))
p1 = (sp.Integer(-1), sp.Integer(4))

# Center lies on the line: 2h + 3k = 3
eq_line = sp.Eq(2*h + 3*k, 3)

# Equidistant to both points (perpendicular bisector condition):
eq_eqdist = sp.Eq((p0[0] - h)**2 + (p0[1] - k)**2,
                  (p1[0] - h)**2 + (p1[1] - k)**2)

# Solve for center (h, k)
sol = sp.solve([eq_line, eq_eqdist], (h, k), dict=True)
center = (sp.simplify(sol[0][h]), sp.simplify(sol[0][k]))

# Radius squared from center to either point
r2 = sp.simplify((p0[0] - center[0])**2 + (p0[1] - center[1])**2)

# Standard equation: (x - h)^2 + (y - k)^2 = r^2
standard_eq = sp.Eq((x - center[0])**2 + (y - center[1])**2, r2)

print("Center:", center)
print("r^2:", r2)
print("Standard:", standard_eq)

# Verify both points lie on the circle
ok_p0 = sp.simplify(standard_eq.lhs.subs({x: p0[0], y: p0[1]})) == r2
ok_p1 = sp.simplify(standard_eq.lhs.subs({x: p1[0], y: p1[1]})) == r2
print("Passes through (8,-5):", ok_p0)
print("Passes through (-1,4):", ok_p1)
```

```python
# 15 Find the standard equation of the circle with center (3, 5) that is tangent to the line 12x -5y +2 = 0.

import sympy as sp

x, y = sp.symbols("x y", real=True)

# Given
h, k = sp.Integer(3), sp.Integer(5)
A, B, C = sp.Integer(12), sp.Integer(-5), sp.Integer(2)  # 12x - 5y + 2 = 0

# Radius = distance from center to line
r = sp.Abs(A * h + B * k + C) / sp.sqrt(A**2 + B**2)
r2 = sp.simplify(r**2)

standard_eq = sp.Eq((x - h) ** 2 + (y - k) ** 2, r2)

print("r  =", sp.simplify(r))
print("r^2=", r2)
print(f"Standard equation: {standard_eq}")
```
