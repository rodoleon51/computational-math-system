### 1. Mathematical Background

**Distance formula**
$$d = \sqrt{ (x_2 - x_1)^2 + (y_2 - y_1)^2 }$$

**Midpoint formula**
$$M = \left( \frac{x_1 + x_2}{2} , \frac{y_1 + y_2}{2} \right)$$

**Slope formula**
$$m = \frac{y_2 - y_1}{x_2 - x_1}$$

**Point–slope form of a line**
$$y - y_1 = m(x - x_1)$$

**Slope–intercept form of a line**
$$y = mx + b$$

**Distance from a point to a line**
$$Ax + By + C = 0$$
$$d = \frac{|Ax_1 + By_1 + C|}{\sqrt{A^2 + B^2}}$$

**Quadrants**
- $Q1: x>0, y>0$
- $Q2: x<0, y>0$
- $Q3: x<0, y<0$
- $Q4: x>0, y<0$

```python
# Ch2_2. Is the triangle with vertices A(1,5), B(4,2), and C(5,6) isosceles?
A = (1, 5)
B = (4, 2)
C = (5, 6)
distance_AB = distance(A, B)
print(f"Distance AB: {distance_AB}")
distance_BC = distance(B, C)
print(f"Distance BC: {distance_BC}")
distance_CA = distance(C, A)
print(f"Distance CA: {distance_CA}")
isosceles = distance_AB == distance_BC or distance_BC == distance_CA or distance_CA == distance_AB
print(f"Triangle ABC is isosceles: {isosceles}")
```

```python
#Ch2_3. Is the triangle with these vertices a right triangle?
A = (-5, 6)
B = (2, 3)
C = (5, 10)
distance_AB = distance(A, B)
print(f"Distance AB: {distance_AB}")
distance_BC = distance(B, C)
print(f"Distance BC: {distance_BC}")
distance_CA = distance(C, A)
print(f"Distance CA: {distance_CA}")
distances = sorted([distance_AB, distance_BC, distance_CA])
right_triangle = sp.simplify(distances[0]**2 + distances[1]**2 - distances[2]**2) == 0
print(f"Triangle ABC is right triangle: {right_triangle}")
# 58 + 58 = 116, hence it is a right triangle
```

```python
# Ch2_5 Find the coordinates of the point that divides the line segment joining the points A(1,2) and B(6,7) in the ratio 2:3.
A = (1, 2)
B = (6, 7)
ratio = (2, 3)
# Find the 2:3 ratio in the x and y coordinates of AB
x = (ratio[0] * B[0] + ratio[1] * A[0]) / (ratio[0] + ratio[1])
y = (ratio[0] * B[1] + ratio[1] * A[1]) / (ratio[0] + ratio[1])
point_dividing = (x, y)
print(f"Point dividing AB in ratio 2:3 is: {point_dividing}")
```

```python
# Ch2_Suplementary 9. Find the area of triangle with vertices A(2, 5), B(2, -5), and C(-3, 5).
A = (2, 5)
B = (2, -5)
C = (-3, 5)
base = distance(A, C)
height = distance(A, B)
area = 0.5 * base * height
print(f"Area of the triangle is: {area:.0f}")
```

```python
# Ch2_Supplementqry 13a. If rectangle find the area.
A = (10, 6)
B = (3, 3)
C = (6, -4)
distance_AB = distance(A, B)
print(f"Distance AB: {distance_AB}")
distance_BC = distance(B, C)
print(f"Distance BC: {distance_BC}")
distance_CA = distance(C, A)
print(f"Distance CA: {distance_CA}")
distances = sorted([distance_AB, distance_BC, distance_CA])
right_triangle = (
    sp.simplify(distances[0] ** 2 + distances[1] ** 2 - distances[2] ** 2) == 0
)
print(f"Triangle ABC is right triangle: {right_triangle}")
if right_triangle:
    area = 0.5 * distances[0] * distances[1]
    print(f"Area of the triangle is: {area:.0f}")
```
